# MetaSignal stdpy Python Implementation

This notebook demonstrates the use of the `metasignal.stdpy` library, which is a pure Python port of the MATLAB Signal Detection Theory (SDT) code.

```python
import numpy as np
from metasignal import stdpy
import matplotlib.pyplot as plt
```

## Generate Sample Data

```python
# 100 trials, 4 ratings
n_trials = 100
n_ratings = 4

stim = np.random.choice([0, 1], n_trials)
resp = stim.copy()
# Add some noise to responses
noise_mask = np.random.rand(n_trials) < 0.2
resp[noise_mask] = 1 - resp[noise_mask]

conf = np.random.randint(1, n_ratings + 1, n_trials)
# High confidence for correct trials usually
correct = (stim == resp)
conf[correct] = np.random.randint(3, 5, np.sum(correct))
conf[~correct] = np.random.randint(1, 3, np.sum(~correct))
```

## Basic SDT Measures

```python
dprime, c, ln_beta = stdpy.compute_sdt_resp(stim, resp)
print(f"d': {dprime:.3f}")
print(f"c: {c:.3f}")
print(f"ln(beta): {ln_beta:.3f}")
```

## Meta-d' (MLE Estimation)

```python
# Convert trials to counts first
nR_S1, nR_S2 = stdpy.trials_to_counts(stim, resp, conf, n_ratings)
print(f"Counts S1: {nR_S1}")
print(f"Counts S2: {nR_S2}")

res = stdpy.fit_meta_d_mle(nR_S1, nR_S2)
print(f"Meta-da: {res['meta_da']:.3f}")
print(f"M-Ratio: {res['M_ratio']:.3f}")
```

## Other Meta-Measures

```python
auc = stdpy.compute_type2_auc(nR_S1, nR_S2)
gamma = stdpy.compute_gamma(nR_S1, nR_S2)
phi = stdpy.compute_phi(nR_S1, nR_S2)

print(f"Type-2 AUC: {auc:.3f}")
print(f"Gamma: {gamma:.3f}")
print(f"Phi: {phi:.3f}")
```
